Algorithm: Logistic Regression using Gradient Descent

1. Create the training dataset X with two input features and the output values y.

2. Set the number of training examples m.

3. Initialize the parameters:
   w1 = 0
   w2 = 0
   b = 0

4. Set the learning rate alpha.

5. Define the Sigmoid function:
   sigmoid(z) = 1 / (1 + e^(-z))

6. Repeat the following steps for a fixed number of iterations:

   a. Initialize:
      dw1 = 0
      dw2 = 0
      db = 0

   b. For each training example:
        i. Calculate the linear equation:
         z = w1*x1 + w2*x2 + b
         
        ii. Calculate the prediction using the Sigmoid function:
          prediction = sigmoid(z)
          
        iii. Calculate the error:
           error = prediction - y

        iv. Calculate the gradients:
          dw1 = dw1 + error*x1
          dw2 = dw2 + error*x2
          db = db + error

   c. Calculate the average gradients:
      dw1 = dw1 / m
      dw2 = dw2 / m
      db = db / m

   d. Update the parameters:
      w1 = w1 - alpha*dw1
      w2 = w2 - alpha*dw2
      b = b - alpha*db

7. Print the final values of w1, w2 and b.

8. For a new input (x1, x2), calculate:
   z = w1*x1 + w2*x2 + b

9. Calculate the probability:
   probability = sigmoid(z)

10. If probability >= 0.5:
       Predict class 1
    Otherwise:
       Predict class 0

11. Print the predicted class and probability.

```python
# Logistic Regression

X = [
    [2, 30],
    [3, 40],
    [4, 50],
    [5, 60],
    [6, 70]
]

y = [0, 0, 0, 1, 1]

m = 5

# Initialize parameters
w1 = 0
w2 = 0
b = 0

alpha = 0.01

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + 2.71828 ** (-z))


# Gradient Descent
for iteration in range(10000):

    dw1 = 0
    dw2 = 0
    db = 0

    for i in range(m):

        # Linear equation
        z = w1 * X[i][0] + w2 * X[i][1] + b

        # Sigmoid
        prediction = sigmoid(z)

        error = prediction - y[i]

        dw1 = dw1 + error * X[i][0]
        dw2 = dw2 + error * X[i][1]
        db = db + error

    dw1 = dw1 / m
    dw2 = dw2 / m
    db = db / m

    # Update parameters
    w1 = w1 - alpha * dw1
    w2 = w2 - alpha * dw2
    b = b - alpha * db


print("w1 =", w1)
print("w2 =", w2)
print("b =", b)


# Prediction
x1 = 5
x2 = 65

z = w1 * x1 + w2 * x2 + b

probability = sigmoid(z)

if probability >= 0.5:
    print("Prediction: 1")
else:
    print("Prediction: 0")

print("Probability =", probability)
```

```python
import numpy as np

# -----------------------------
# 1. Training data
# -----------------------------
X = np.array([
    [2, 60],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85],
    [8, 90],
    [9, 95]
], dtype=float)

y = np.array([0, 0, 0, 1, 1, 1, 1, 1], dtype=float)

# -----------------------------
# 2. Initialize parameters
# -----------------------------
theta0 = 0.0
theta1 = 0.0
theta2 = 0.0

learning_rate = 0.01
iterations = 1000

m = len(y)

# -----------------------------
# 3. Sigmoid function
# -----------------------------
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# -----------------------------
# 4. Gradient Descent
# -----------------------------
for i in range(iterations):

    # Calculate z
    z = theta0 + theta1 * X[:, 0] + theta2 * X[:, 1]

    # Calculate prediction
    y_pred = sigmoid(z)

    # Calculate gradients
    dtheta0 = (1/m) * np.sum(y_pred - y)

    dtheta1 = (1/m) * np.sum((y_pred - y) * X[:, 0])

    dtheta2 = (1/m) * np.sum((y_pred - y) * X[:, 1])

    # Update parameters
    theta0 = theta0 - learning_rate * dtheta0
    theta1 = theta1 - learning_rate * dtheta1
    theta2 = theta2 - learning_rate * dtheta2

# -----------------------------
# 5. Print learned parameters
# -----------------------------
print("Theta 0:", theta0)
print("Theta 1:", theta1)
print("Theta 2:", theta2)

# -----------------------------
# 6. Predict for new student
# -----------------------------
x1 = 6
x2 = 78

z = theta0 + theta1 * x1 + theta2 * x2

probability = sigmoid(z)

print("Probability of passing:", probability)

# -----------------------------
# 7. Apply threshold
# -----------------------------
if probability >= 0.5:
    print("Prediction: Pass")
else:
    print("Prediction: Fail")
```
