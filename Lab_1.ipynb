```python
import numpy as np
#Linear Regression using Gradient Descent.
X = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 4],
    [2500, 4],
    [3000, 5]
], dtype=float)

y = np.array([200, 300, 400, 500, 600], dtype=float)

# Normalize X
X = (X - np.mean(X, axis=0)) / np.std(X, axis=0)

m = len(y)

w1 = 0
w2 = 0
b = 0

alpha = 0.01
iterations = 10000

for i in range(iterations):

    y_pred = b + w1 * X[:, 0] + w2 * X[:, 1]

    error = y_pred - y

    cost = (1 / (2 * m)) * np.sum(error ** 2)

    dw1 = (1 / m) * np.sum(error * X[:, 0])
    dw2 = (1 / m) * np.sum(error * X[:, 1])
    db = (1 / m) * np.sum(error)

    w1 = w1 - alpha * dw1
    w2 = w2 - alpha * dw2
    b = b - alpha * db

    if i % 1000 == 0:
        print("Iteration:", i, "Cost:", cost)

print("\nFinal Parameters:")
print("w1 =", w1)
print("w2 =", w2)
print("b =", b)
```

```python
# Multiple Linear Regression
X = [
    [1000, 2],
    [1500, 3],
    [2000, 4],
    [2500, 4],
    [3000, 5]
]

y = [200, 300, 400, 500, 600]

m = 5

# Initialize parameters
w1 = 0
w2 = 0
b = 0

alpha = 0.0000001

# Gradient Descent
for iteration in range(10000):

    dw1 = 0
    dw2 = 0
    db = 0

    for i in range(m):

        prediction = w1 * X[i][0] + w2 * X[i][1] + b

        error = prediction - y[i]

        dw1 = dw1 + error * X[i][0]
        dw2 = dw2 + error * X[i][1]
        db = db + error

    dw1 = dw1 / m
    dw2 = dw2 / m
    db = db / m

    w1 = w1 - alpha * dw1
    w2 = w2 - alpha * dw2
    b = b - alpha * db

print("w1 =", w1)
print("w2 =", w2)
print("b =", b)

# Prediction
area = 2000
bedrooms = 4

prediction = w1 * area + w2 * bedrooms + b

print("Predicted price =", prediction)
```
