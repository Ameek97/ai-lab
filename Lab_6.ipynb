# LAB 06: Single-Layer Neural Network for Binary Classification

## Aim

To implement a single-layer neural network (perceptron) for binary classification of the logical AND dataset.

## Objectives

- Represent binary inputs and target labels.
- Initialize perceptron weights and bias.
- Calculate a weighted sum and apply a step activation.
- Update parameters using the perceptron learning rule.
- Test the trained perceptron on all input combinations.

## Algorithm / Steps

1. Define the four two-feature input vectors and the AND targets.
2. Initialize the weight vector and bias to zero, then set the learning rate and number of epochs.
3. For each training example, compute
   \[
   z = \mathbf{x}\cdot\mathbf{w}+b
   \]
   and apply the step function, returning 1 when \(z>0\), otherwise 0.
4. Calculate the error \(e=y-\hat{y}\).
5. Update the parameters using
   \[
   \mathbf{w}\leftarrow\mathbf{w}+lpha e\mathbf{x}, \qquad
   b\leftarrow b+lpha e.
   \]
6. Repeat for all configured epochs and then predict every input vector.

LAB 06: Implement a Single-Layer Neural Network for Classification

Aim

To implement a Single-Layer Neural Network (Perceptron) for binary classification using Python.

Objectives

Understand the basic structure of a single-layer neural network.

Initialize weights and bias.

Calculate the weighted sum.

Apply an activation function.

Update weights and bias using the perceptron learning rule.

Classify input data into two classes.

## Source Code

The original implementation cells follow in their existing order.

```python
import numpy as np

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target output for AND gate
y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

# Learning rate
learning_rate = 0.1

# Number of epochs
epochs = 10


# Step activation function
def activation(z):
    if z > 0:
        return 1
    else:
        return 0


# Training
for epoch in range(epochs):

    print("Epoch:", epoch + 1)

    for i in range(len(X)):

        # Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Predict output
        prediction = activation(z)

        # Calculate error
        error = y[i] - prediction

        # Update weights
        weights = weights + learning_rate * error * X[i]

        # Update bias
        bias = bias + learning_rate * error

    print("Weights:", weights)
    print("Bias:", bias)


# Testing
print("\nFinal Predictions:")

for i in range(len(X)):

    z = np.dot(X[i], weights) + bias

    prediction = activation(z)

    print(
        "Input:", X[i],
        "Actual:", y[i],
        "Predicted:", prediction
    )
```
