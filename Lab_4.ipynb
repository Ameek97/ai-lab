## Algorithm / Steps

### Part A: 8-Tile Puzzle with BFS

1. Encode each board as a tuple and define the goal tuple.
2. Place the initial state in a queue and mark it visited.
3. Remove a state, locate the blank tile, and generate each legal swap.
4. Record each new state's parent and enqueue it once.
5. When the goal is reached, follow parent links, reverse the path, and print each board.

### Part B: Tic-Tac-Toe with Minimax

1. Represent the board as nine positions and define all winning triples.
2. Return \(1\) for an O win, \(-1\) for an X win, and \(0\) for a draw.
3. Recursively try every legal move: O maximizes the score and X minimizes it.
4. For each computer turn, evaluate every legal O move and select the move with the highest score.
5. Continue alternating human input and computer moves until a result is reached.

### Part C: 8-Queens with Backtracking

1. Start with an empty \(8 	imes 8\) board represented by one column per row.
2. Try each column in the current row.
3. Reject a column if it shares a column or diagonal with a previously placed queen.
4. Place a safe queen and recursively solve the next row.
5. If no column works, remove the previous queen and backtrack; report the board when all rows are filled.

8 Tiles using BFS:
Breadth-First Search(BFS) explores states level by level, examining all possible moves at the current depth before moving deeper.
1. Explores states level by level.
2. Guarantees the shortest solution path.
3. Suitable for finding optimal solutions.
4. Requires high memory for large search spaces.

## Source Code

The original implementation cells follow in their existing order.

```python
from collections import deque
# 8 Tiles
# Goal state
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

# Possible movements of blank space
moves = {
    0: [1, 3],
    1: [0, 2, 4],
    2: [1, 5],
    3: [0, 4, 6],
    4: [1, 3, 5, 7],
    5: [2, 4, 8],
    6: [3, 7],
    7: [4, 6, 8],
    8: [5, 7]
}


def bfs(start):

    queue = deque()
    queue.append(start)

    visited = set()
    visited.add(start)

    parent = {}

    while queue:

        current = queue.popleft()

        # Goal reached
        if current == goal:
            path = []

            while current != start:
                path.append(current)
                current = parent[current]

            path.append(start)
            path.reverse()

            return path

        # Position of blank
        blank = current.index(0)

        # Generate next states
        for position in moves[blank]:

            new_state = list(current)

            # Swap blank with tile
            new_state[blank], new_state[position] = \
                new_state[position], new_state[blank]

            new_state = tuple(new_state)

            if new_state not in visited:

                visited.add(new_state)
                queue.append(new_state)

                parent[new_state] = current

    return None


# Initial state
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

# Solve
solution = bfs(start)


# Print solution
if solution:

    print("Solution found!")
    print("Number of moves:", len(solution) - 1)

    for state in solution:

        print()

        print(state[0], state[1], state[2])
        print(state[3], state[4], state[5])
        print(state[6], state[7], state[8])

else:
    print("No solution found")
```

Tic Tac Toe:
For Tic-Tac-Toe, we can represent every board configuration as a state and search possible moves.
The following implementation uses Minimax, which is a game-playing search technique.

```python
import math

board = [" "] * 9


def print_board():

    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner():

    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:

        if board[a] == board[b] == board[c] != " ":
            return board[a]

    if " " not in board:
        return "Draw"

    return None


def minimax(is_maximizing):

    result = check_winner()

    if result == "O":
        return 1

    if result == "X":
        return -1

    if result == "Draw":
        return 0

    if is_maximizing:

        best_score = -math.inf

        for i in range(9):

            if board[i] == " ":

                board[i] = "O"

                score = minimax(False)

                board[i] = " "

                best_score = max(best_score, score)

        return best_score

    else:

        best_score = math.inf

        for i in range(9):

            if board[i] == " ":

                board[i] = "X"

                score = minimax(True)

                board[i] = " "

                best_score = min(best_score, score)

        return best_score


def best_move():

    best_score = -math.inf
    move = None

    for i in range(9):

        if board[i] == " ":

            board[i] = "O"

            score = minimax(False)

            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move


print("TIC-TAC-TOE")
print("Positions are 0 to 8")

print_board()

while True:

    # Human move
    position = int(input("Enter your move (0-8): "))

    if board[position] != " ":
        print("Position already occupied!")
        continue

    board[position] = "X"

    print_board()

    result = check_winner()

    if result:
        print("Result:", result)
        break

    # Computer move
    computer_move = best_move()
    board[computer_move] = "O"

    print("Computer selected:", computer_move)

    print_board()

    result = check_winner()

    if result:
        print("Result:", result)
        break
```

8-Queens Problem using Backtracking Algorithm
Start with an empty chessboard.
Place one queen in each row.
Check whether the selected column is safe.
A queen cannot share:
the same column,
the same diagonal.
If the position is safe, place the queen.
If no safe position exists, backtrack.
Continue until all 8 queens are placed.

```python
N = 8

board = [-1] * N


def is_safe(row, col):

    for previous_row in range(row):

        previous_col = board[previous_row]

        # Same column
        if previous_col == col:
            return False

        # Same diagonal
        if abs(previous_row - row) == abs(previous_col - col):
            return False

    return True


def solve(row):

    if row == N:
        return True

    for col in range(N):

        if is_safe(row, col):

            board[row] = col

            if solve(row + 1):
                return True

            # Backtracking
            board[row] = -1

    return False


def print_board():

    for row in range(N):

        for col in range(N):

            if board[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()


if solve(0):
    print("Solution for 8-Queens:")
    print_board()
else:
    print("No solution exists.")
```
