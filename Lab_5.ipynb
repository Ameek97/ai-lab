# LAB 05: Minimax Algorithm for Game Playing

## Aim

To implement the Minimax algorithm for a two-player game tree and determine the optimal value for the maximizing player.

## Objectives

- Represent terminal game outcomes as utility values.
- Traverse a binary game tree recursively.
- Distinguish MAX and MIN decision levels.
- Select the maximum or minimum child value at each level.
- Determine the optimal value returned from the root state.

## Algorithm / Steps

1. Store the utility values of the terminal nodes and define the maximum tree depth.
2. Begin at the root with the maximizing player.
3. If the current depth equals the maximum depth, return the corresponding terminal utility.
4. At a MAX node, recursively evaluate both children and return their maximum.
5. At a MIN node, recursively evaluate both children and return their minimum.
6. Continue until the root receives the optimal value and print it.

```python
LAB 05: Implement the Minimax Algorithm for Game Playing

Aim

To implement the Minimax algorithm for game playing and determine the optimal move for a player by exploring the game tree.

Objectives

Understand game-tree representation.

Implement the Minimax decision-making strategy.

Identify MAX and MIN players.

Select the optimal move based on the game state.


Algorithm

1. Represent the current game as a state.


2. Generate all possible moves.


3. Create a game tree by recursively exploring possible moves.


4. Assign a utility value to terminal states.


5. MAX player selects the maximum value.


6. MIN player selects the minimum value.


7. Continue until the optimal move is obtained.
```

```python
import math

def minimax(depth, node_index, maximizing_player, values, max_depth):

    # Terminal node
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        return max(
            minimax(depth + 1, node_index * 2, False,
                    values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False,
                    values, max_depth)
        )

    else:
        return min(
            minimax(depth + 1, node_index * 2, True,
                    values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True,
                    values, max_depth)
        )


# Utility values of terminal states
values = [3, 5, 2, 9, 12, 5, 23, 23]

max_depth = 3

optimal_value = minimax(
    0, 0, True, values, max_depth
)

print("Optimal value:", optimal_value)
```
