LAB 07: Implement Forward Chaining and Backward Chaining Algorithms

Aim
To implement Forward Chaining and Backward Chaining algorithms in Artificial Intelligence to perform logical reasoning
and infer conclusions from a set of facts and rules.

Understand rule-based reasoning in AI.

Objectives
✓Implement forward chaining using facts and inference rules.
✓Implement backward chaining to prove a given goal.
✓Compare data-driven and goal-driven reasoning.

✓✓Part A: Forward Chaining

Forward Chaining is a data-driven inference technique that starts with known facts and repeatedly applies
 rules to derive new facts until the goal is reached or no more rules can be applied.

Example rules:
Rule 1: If A and B are true, then C is true.
Rule 2: If C is true, then D is true.
Facts: A, B
Conclusion: C, then D.


✓✓Part B: Backward Chaining

Backward Chaining is a goal-driven inference technique that starts with a goal and works backward to determine whether the known facts and rules support it.

Using the same rules:
Rule 1: If A and B are true, then C is true.
Rule 2: If C is true, then D is true.
Rule 3: If D is true, then E is true.
Goal: E
The algorithm works backward from E to D, then C, and finally verifies A and B.





Difference between Forward and Backward Chaining
Forward Chaining	                                                 Backward Chaining
Data-driven reasoning	                                          Goal-driven reasoning
Starts with known facts	                                          Starts with a goal
Applies rules to derive conclusions	                              Searches for rules that support the goal
Stops when the goal is reached or no new facts can be derived	  Stops when the goal is proved or cannot be proved
Example: production systems	                                      Example: Prolog-style reasoning

```python
# Forward Chaining

facts = ["A", "B"]

rules = [
    (["A", "B"], "C"),
    (["C"], "D"),
    (["D"], "E")
]
# Each rule is a tuple containing two parts:
#   - condition: the facts that must be true.
#   - conclusion: the new fact we can infer if the conditions are satisfied.

goal = "E"
# We want to find out whether E can be derived from the facts and rules.


# while True keeps repeating the process.
# new_fact_added = False assumes that no new fact has been discovered yet in the current iteration.


while True:
    new_fact_added = False

# This loop visits every rule, one by one.
    for condition, conclusion in rules:
        if all(fact in facts for fact in condition):
            if conclusion not in facts:
                facts.append(conclusion)
                print("Derived:", conclusion)
                new_fact_added = True

    if goal in facts or not new_fact_added:
        break


# fact in facts checks whether a particular fact exists in our knowledge base.
# for fact in condition checks each required fact.
# all(...) returns True only if every condition is satisfied.
if goal in facts:
    print("Goal", goal, "is reached")
else:
    print("Goal", goal, "cannot be reached")

print("Final facts:", facts)
```

```python
# Backward Chaining

facts = ["A", "B"]
# This is our initial knowledge base. We already know that A and B are true.

rules = [
    (["A", "B"], "C"),
    (["C"], "D"),
    (["D"], "E")
]

def backward_chaining(goal):
    if goal in facts:
        return True

    for condition, conclusion in rules:
        if conclusion == goal:
            if all(backward_chaining(fact) for fact in condition):
                return True

    return False

goal = "E"

if backward_chaining(goal):
    print("Goal", goal, "can be proved")
else:
    print("Goal", goal, "cannot be proved")
```
