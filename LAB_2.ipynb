```python
#Python Implementation of DFS


def dfs(graph, start, goal):

    stack = [start]
    visited = set()

    while stack:

        node = stack.pop()

        if node in visited:
            continue

        print("Visiting:", node)
        visited.add(node)

        if node == goal:
            print("Goal found!")
            return

        for neighbour in reversed(graph[node]):
            if neighbour not in visited:
                stack.append(neighbour)

    print("Goal not found")


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

dfs(graph, 'A', 'F')
```

```python
#Breadth First Search (BFS)

from collections import deque

def bfs(graph, start, goal):

    queue = deque([start])
    visited = set()

    while queue:

        node = queue.popleft()

        if node in visited:
            continue

        print("Visiting:", node)
        visited.add(node)

        if node == goal:
            print("Goal found!")
            return

        for neighbour in graph[node]:
            if neighbour not in visited:
                queue.append(neighbour)

    print("Goal not found")


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

bfs(graph, 'A', 'F')
```

Algorithm: A* Search
1. Create a graph using a dictionary and store the cost/weight of each edge.
2. Assign a heuristic value to each node of the graph.
3. Define the Start node and Goal node.
4. Create an Open List and a Closed List.
   Put the Start node in the Open List.
   Set the parent of the Start node as None.
   Set the cost g of the Start node as 0.
5. Repeat while the Open List is not empty:
   a. Select the node from the Open List having the minimum
      f(n) = g(n) + h(n)

      where:
      g(n) = actual cost from Start node to current node
      h(n) = heuristic cost from current node to Goal node

   b. If the selected node is the Goal node:
      i. Trace the path using the parent of each node.
      ii. Reverse the path.
      iii. Print the path and total cost.
      iv. Stop.

   c. Remove the selected node from the Open List
      and add it to the Closed List.

   d. Check all the neighbouring nodes of the current node.

   e. If a neighbouring node is already in the Closed List,
      skip that node.

   f. Calculate the new cost:
      new_g = g(current) + cost

   g. If the neighbouring node is not in the Open List:
      i. Add it to the Open List.
      ii. Set its g value to new_g.
      iii. Set its parent as the current node.

   h. Otherwise, if new_g is less than the existing g value:
      i. Update its g value.
      ii. Update its parent.
6. Continue until the Goal node is reached or the Open List becomes empty.
7. Display the final path and its total cost.

```python
# A* Search

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('E', 1)],
    'D': [('G', 3)],
    'E': [('G', 2)],
    'G': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 4,
    'D': 3,
    'E': 2,
    'G': 0
}

start = 'A'
goal = 'G'

open_list = [start]
closed_list = []

g = {}
parent = {}

g[start] = 0
parent[start] = None

while len(open_list) > 0:

    # Find node with minimum f = g + h
    current = open_list[0]

    for node in open_list:
        f1 = g[current] + heuristic[current]
        f2 = g[node] + heuristic[node]

        if f2 < f1:
            current = node

    # Goal reached
    if current == goal:

        path = []

        while current is not None:
            path.append(current)
            current = parent[current]

        path.reverse()

        print("Path:", path)
        print("Cost:", g[goal])

        break

    open_list.remove(current)
    closed_list.append(current)

    # Check neighbours
    for neighbour, cost in graph[current]:

        if neighbour in closed_list:
            continue

        new_g = g[current] + cost

        if neighbour not in open_list:
            open_list.append(neighbour)

            g[neighbour] = new_g
            parent[neighbour] = current

        elif new_g < g[neighbour]:
            g[neighbour] = new_g
            parent[neighbour] = current
```

Algorithm: Hill Climbing
1. Start
2. Define the set of values/states.
3. Set the current state as the starting state.
4. Find the neighboring state of the current state.
5. Compare the neighbor with the current state.
6. If the neighbor is better than the current state:

   * Move to the neighbor.
   * Make it the current state.
7. Otherwise, stop.
8. Display the current state as the maximum/best state.
9. End.


```python
# Hill Climbing Algorithm

a = [2, 5, 8, 12, 10, 7, 4]

current = 0

while True:
    next = current + 1

    if next < len(a) and a[next] > a[current]:
        current = next
    else:
        break

print("Maximum value =", a[current])
print("Position =", current)
```
